# Preprocessing the Dataset

### 1. Import Required Libraries

```python
import pandas as pd
from datetime import datetime
from sklearn import preprocessing
from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder
```

### 2. Load Dataset

```python
df = pd.read_csv("heart_disease_uci.csv")

# Display all column names
print("Original Columns:")
print(df.columns.tolist())
```

### 3. Select Required Features

```python
selected_features = [
    'age',
    'sex',
    'cp',
    'trestbps',
    'fbs',
    'restecg',
    'thalch',
    'exang',
    'oldpeak',
    'slope',
    'num'
]

df = df[selected_features]
```

### 4. Rename 'num' to 'target'

```python
df.rename(columns={'num': 'target'}, inplace=True)
```

### 5. Display Dataset Information

```python
print("\nDataset Shape:", df.shape)

print("\nColumns:")
print(df.columns.tolist())

print("\nFirst Five Rows:")
print(df.head())
```

### 6. Preprocessing features

- Missing Values Handling

```python
# Remove Rows with Missing Values

print("Dataset shape before removing missing values:", df.shape)

# Remove all rows containing one or more missing values
df.dropna(inplace=True)

# Reset the index
df.reset_index(drop=True, inplace=True)

print("Dataset shape after removing missing values:", df.shape)

print("\nMissing values after removal:")
print(df.isnull().sum())
```

- convert target value more than 1 as disease = 1 else no disease = 0

```python
# Convert target to binary (0 = No Heart Disease, 1 = Heart Disease)
df['target'] = (df['target'] > 0).astype(int)
```

- Find the number of disease and undiseased cases

```python
labels = (df['target'] > 0)
features = df.drop(['target'], axis = 1)

print(f'Disease cases: {sum(labels == 1):8d}')
print(f'Healthy cases: {sum(labels == 0):8d}')
```

```python
print(df.head())
```

- Separate the features according to the type

```python
bin_features = ['sex', 'fbs', 'exang']
num_features = ['age', 'trestbps', 'thalch', 'oldpeak']
cat_features = ['cp', 'restecg', 'slope']
```

- Convert Binary Categorical Features to Numeric

```python
# Remove leading/trailing spaces
df['sex'] = df['sex'].astype(str).str.strip()
df['fbs'] = df['fbs'].astype(str).str.strip()
df['exang'] = df['exang'].astype(str).str.strip()

# Convert to numeric
df['sex'] = df['sex'].replace({
    'Male': 1,
    'Female': 0
})

df['fbs'] = df['fbs'].replace({
    'True': 1,
    'False': 0,
})

df['exang'] = df['exang'].replace({
    'True': 1,
    'False': 0,
})

# Convert columns to integer
df['sex'] = df['sex'].astype(int)
df['fbs'] = df['fbs'].astype(int)
df['exang'] = df['exang'].astype(int)

encoded_bin_features = pd.concat([df['sex'], df['fbs'],df['exang']], axis=1)
# Display unique values after conversion
print("\nUnique values after conversion:")
print(encoded_bin_features)
```

```python
print(features.columns.tolist())
```

- Use minmax scaler to preprocess numerical features

```python
transformer = preprocessing.RobustScaler()

num_values = pd.DataFrame(
    transformer.fit_transform(features[num_features]),
    columns=num_features,
    index=features.index
)

print(num_values.head())
```

- preprocessing categorical features

```python
# One-Hot Encode cp and restecg

onehot_encoder = OneHotEncoder(sparse_output=False, handle_unknown='ignore')

onehot_encoded = onehot_encoder.fit_transform(df[['cp', 'restecg']])

onehot_df = pd.DataFrame(
    onehot_encoded,
    columns=onehot_encoder.get_feature_names_out(['cp', 'restecg']),
    index=df.index
)

# Ordinal Encode slope
ordinal_encoder = OrdinalEncoder()

ordinal_encoded = ordinal_encoder.fit_transform(df[['slope']])

ordinal_df = pd.DataFrame(
    ordinal_encoded,
    columns=['slope'],
    index=df.index
)

# Encoded categorical features
encoded_cat_features = pd.concat([onehot_df, ordinal_df], axis=1)

# Display first 5 rows
print(encoded_cat_features)
```

- completely preprocessed dataset

```python
preprocessed_features = pd.concat([encoded_bin_features,num_values, encoded_cat_features, df['target']], axis=1)
print(preprocessed_features.head())
print(preprocessed_features.columns)
```

- Find the number of disease and undiseased cases

```python
labels = df['target']
features = df.drop(['target'], axis = 1)

print(f'Disease cases: {sum(labels == 1):8d}')
print(f'Healthy cases: {sum(labels == 0):8d}')
```

### 7. Save preprocessed Dataset with Current Date & Time

```python
# Get current date and time
current_time = datetime.now().strftime("%Y-%m-%d_%H-%M-%S")

# Create filename
filename = f"preprocessed_dataset_{current_time}.csv"

# Save DataFrame
preprocessed_features.to_csv(filename, index=False)

print(f"Dataset saved successfully as: {filename}")
```
