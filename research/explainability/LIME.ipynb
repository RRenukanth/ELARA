```python
def positive_contributors(explanation):
    contributors = explanation.as_list()
    positive = []
    for feature,value in contributors:
        if value > 0:
            positive.append((feature,value))

    return positive
```

```python
# Algorithm 3 Encoder Layer

# Feature Encoder
def encode_features(patient_dictionary):
    """ Convert questionnaire information into RL state representation."""

    # Encoding Dictionaries
    sex_dictionary={"Male":1, "Female":0}
    chest_pain_dictionary={"Typical Angina":0, "Atypical Angina":1, "Non Anginal Pain":2, "Asymptomatic":3}
    fbs_dictionary={"True":1, "False":0}
    restecg_dictionary={"Normal":0, "ST-T Wave Abnormality":1, "Left Ventricular Hypertrophy":2}
    exang_dictionary={"Yes":1, "No":0}
    slope_dictionary={"Upsloping":0, "Flat":1, "Down Sloping":2}

    # Numerical Features
    age = float(patient_dictionary["Age"])
    trestbps = float(patient_dictionary["Resting Blood Pressure"])
    thalach = float(patient_dictionary["Maximum Heart Rate"])
    oldpeak = float(patient_dictionary["Old Peak"])

    # Categorical Features
    sex = sex_dictionary[patient_dictionary["Sex"]]
    cp = chest_pain_dictionary[patient_dictionary["Chest Pain"]]
    fbs = fbs_dictionary[patient_dictionary["FBS"]]
    restecg = restecg_dictionary[patient_dictionary["Rest ECG"]]
    exang = exang_dictionary[patient_dictionary["Exang"]]
    slope = slope_dictionary[patient_dictionary["Slope"]]

    # Arrange Features
    patient_state=[age, sex, cp, trestbps, fbs, restecg, thalach, exang, oldpeak, slope]

    # Generate Numpy Array
    patient_state = np.array(patient_state, dtype=np.float32)

    # Generate Tensor
    patient_tensor = torch.FloatTensor(patient_state)

    # Display Encoded Features
    print("\n")
    print("="*70)
    print("ENCODED PATIENT FEATURES")
    print("="*70)
    print(patient_state)
    print("="*70)

    # Return Results
    return patient_dictionary,\ patient_state,\ encoded_state\ patient_tensor

# Execute Encoder Layer
if __name__=="__main__":
    patient_information={"Age":63,
                         "Sex":"Male",
                         "Chest Pain":"Typical Angina",
                         "Resting Blood Pressure":145,
                         "FBS":"True",
                         "Rest ECG":"Normal",
                         "Maximum Heart Rate":150,
                         "Exang":"Yes",
                         "Old Peak":2.3,
                         "Slope":"Flat"}
    patient_state,\
    patient_tensor=(encode_features(patient_information))
```

```python
# Algorithm 4
# Load Prediction Model
import os
import torch

# Select Prediction Model
def select_prediction_model():
    """ Allow the user to select the prediction model. """

    print("\n")
    print("="*70)
    print("MODEL SELECTION")
    print("="*70)
    print("1. PPO")
    print("2. DQN")
    print("="*70)

    choice = int(input("Select Model : "))

    if choice == 1:
        return "PPO"
    elif choice == 2:
        return "DQN"
    else:
        raise ValueError("Invalid Model Selection.")



# Verify Model File
def verify_model_file(model_path):
    """ Verify whether the saved model exists or not."""
    if not os.path.exists(model_path):
        raise FileNotFoundError(f"{model_path} not found.")

    return True



# Detect Device
def detect_device():
    """ Detect CPU or GPU. """
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    print(f"\nRunning Device : {device}")

    return device


# Load Prediction Model
def load_prediction_model(model_name, model_path, model):
    """ Load the selected model. """
    # Verify Model File
    verify_model_file(model_path)

    # Detect Device
    device = detect_device()

    # Load Saved Parameters
    checkpoint = torch.load(model_path, map_location=device)

    # Load State Dictionary
    model.load_state_dict(checkpoint)
    # Evaluation Mode
    model.eval()

    # Display Information
    print("\n")
    print("="*70)
    print("MODEL INFORMATION")
    print("="*70)
    print(f"Selected Model : {model_name}")
    print(f"Model Loaded Successfully.")
    print(f"Device : {device}")
    print("="*70)

    # Return Results
    return model

# Execute Algorithm 4
if __name__=="__main__":
    model_name = (select_prediction_model())
    print(f"\nSelected Model : {model_name}")
```

```python
# Algorithm 5
# Prediction Layer

import torch
import numpy as np

# Predict Heart Disease
def predict_heart_disease(model, patient_tensor):
    """ Perform heart disease prediction. """
    # Evaluation Mode
    model.eval()

    # Disable Gradient Calculation
    with torch.no_grad():
        # Generate Prediction
        output = model(patient_tensor)

        # Calculate Probabilities
        probabilities = torch.softmax(output, dim=0)

        # Calculate Prediction
        prediction = torch.argmax(probabilities).item()

        # Convert into Percentage
        probability_scores=(probabilities*100).numpy()

        # Prediction Confidence
        prediction_confidence=(torch.max(probabilities).item())*100

        # Predicted Label
        if prediction ==0:
            prediction_label=("No Heart Disease")
        else:
            prediction_label=("Heart Disease")

        # Display Results
        print("\n")
        print("="*70)
        print("PREDICTION RESULTS")
        print("="*70)
        print(f"Prediction : {prediction_label}")
        print(f"\nPrediction Confidence : "f"{prediction_confidence:.2f}%")
        print("\n")
        print(f"No Heart Disease : "f"{probability_scores[0]:.2f}%")
        print(f"Heart Disease : "  f"{probability_scores[1]:.2f}%")
        print("="*70)

        # Return Results
        return(prediction, prediction_label, prediction_confidence, probability_scores, probabilities, output)
```

```python
# Algorithm 6
# Generate LIME Explanation Layer
import numpy as np
from lime.lime_tabular import (LimeTabularExplainer)

# Initialize LIME
def initialize_lime(training_data, feature_names):
    """ Initialize the LIME explainer. """
    explainer = LimeTabularExplainer(training_data = training_data,
                                     feature_names = feature_names,
                                     class_names =["No Heart Disease", "Heart Disease"],
                                     mode = "classification",
                                     discretize_continuous = True)
    return explainer



# Generate LIME Explanation
def generate_lime_explanation(explainer, patient_state, prediction_function):
    """ Generate the LIME explanation. """
    explanation=(explainer.explain_instance(patient_state, prediction_function, num_features=10))

    return explanation


# Extract Feature Contributions
def extract_feature_contributions(explanation):
    """ Extract feature contributions. """
    return explanation.as_list()

# Extract Positive Contributors
def extract_positive_contributors(explanation):
    positive_features=[]

    for feature,weight in (explanation.as_list()):
        if weight >0:
            positive_features.append((feature,weight))
    return positive_features

# Extract Negative Contributors
def extract_negative_contributors(explanation):
    negative_features=[]

    for feature,weight in (explanation.as_list()):
        if weight <0:
            negative_features.append((feature,weight))

    return negative_features

# Calculate Percentage Contribution
def calculate_percentage_contribution(explanation):
    feature_percentages={}
    total_weight = sum(abs(weight) for feature,weight in explanation.as_list())

    for feature,weight in (explanation.as_list()):
        percentage=(abs(weight)/total_weight)*100
        feature_percentages[feature]=round(percentage, 2)

    return feature_percentages



# Rank Features
def rank_features(explanation):
    ranking = sorted(explanation.as_list(), key=lambda x:abs(x[1]), reverse=True)

    return ranking



# Generate LIME Dictionary
def generate_lime_dictionary(explanation):
    lime_dictionary={"Feature Contributions": extract_feature_contributions(explanation),
                     "Positive Contributors": extract_positive_contributors(explanation),
                     "Negative Contributors": extract_negative_contributors(explanation),
                     "Percentage Contributions": calculate_percentage_contribution(explanation),
                     "Feature Ranking": rank_features(explanation)}

    return lime_dictionary
```

```python
# Algorithm 7
# Doctor Explanation Layer

# Generate Clinical Interpretation
def generate_clinical_interpretation(prediction, positive_features, negative_features):
    """ Generate a clinical explanation of the prediction. """
    explanation=""

    # Positive Contributors
    if len(positive_features)!=0:
        explanation +=("\nThe prediction was primarily influenced by the following clinical features:\n")
        for feature,value in (positive_features[:3]):
            explanation +=(f"\n• {feature}")

        explanation +=("\n\nThese features increased the predicted risk of heart disease.")

    # Negative Contributors
    if len(negative_features)!=0:
        explanation +=("\n\nThe following features reduced the predicted risk:\n")

        for feature,value in (negative_features[:3]):
            explanation +=(f"\n• {feature}")

    return explanation


# Clinical Recommendation
def generate_recommendation(prediction):
    """ Generate a recommendation."""

    if prediction ==1:
        return("\nFurther cardiovascular assessment is recommended before confirming the diagnosis.")
    return("\nNo immediate indication of heart disease was identified. However, regular clinical evaluation is recommended.")


# Doctor Explanation Layer
def generate_doctor_explanation(prediction, prediction_label, prediction_confidence, lime_dictionary):
    """ Generate doctor's explanation. """
    positive_features=(lime_dictionary["Positive Contributors"])
    negative_features=(lime_dictionary["Negative Contributors"])
    percentage_scores=(lime_dictionary["Percentage Contributions"])
    feature_ranking=(lime_dictionary["Feature Ranking"])

    print("\n")
    print("="*70)
    print("DOCTOR'S EXPLANATION REPORT")
    print("="*70)


    # Prediction Results
    print(f"\nPrediction : " f"{prediction_label}")
    print(f"\nPrediction Confidence : " f"{prediction_confidence:.2f}%")

    # Positive Contributors
    print("\n")
    print("-"*70)
    print("POSITIVE CONTRIBUTORS")
    print("-"*70)

    for feature,weight in (positive_features):
        percentage=(percentage_scores[feature])
        print(f"{feature}" f" ---> " f"{percentage:.2f}%")

    # Negative Contributors
    print("\n")
    print("-"*70)
    print("NEGATIVE CONTRIBUTORS")
    print("-"*70)

    for feature,weight in (negative_features):
        percentage=(percentage_scores[feature])
        print(f"{feature}" f" ---> " f"{percentage:.2f}%")

    # Feature Ranking
    print("\n")
    print("-"*70)
    print("FEATURE RANKING")
    print("-"*70)

    rank=1
    for feature,weight in (feature_ranking):
        print(f"{rank}) " f"{feature}")
        rank+=1

    # Clinical Interpretation
    interpretation=(generate_clinical_interpretation(prediction, positive_features, negative_features))
    print("\n")
    print("-"*70)
    print("CLINICAL INTERPRETATION")
    print("-"*70)
    print(interpretation)

    # Recommendation
    recommendation=(generate_recommendation(prediction))
    print("\n")
    print("-"*70)
    print("CLINICAL RECOMMENDATION")
    print("-"*70)
    print(recommendation)
    print("\n")
    print("="*70)

    return{"Prediction": prediction_label,
           "Confidence": prediction_confidence,
           "Interpretation": interpretation,
           "Recommendation": recommendation}
```

```python
# Algorithm 8
# Patient Explanation Layer

# Generate Simple Explanation
def generate_simple_explanation(prediction_label):
    """ Generate a simple explanation of the prediction. """

    if prediction_label == ("Heart Disease"):
        return("\nThe AI system predicts that you may have heart disease."
               " Some of your medical measurements have contributed towards this prediction.")
    else:
        return("\nThe AI system predicts that there is currently no indication of"
            " heart disease based on the provided information.")


# Extract Important Features
def extract_important_features(lime_dictionary):
    """Extract the most important positive contributors."""
    positive_features=(lime_dictionary["Positive Contributors"])
    important_features=[]

    for feature,weight in (positive_features[:4]):
        important_features.append(feature)

    return important_features



# Generate Health Recommendation
def generate_health_recommendation(prediction_label):
    """Generate health recommendations."""
    if prediction_label==("Heart Disease"):
        recommendations=["Please consult your physician.",
                         "Regular cardiovascular assessment is recommended.",
                         "Maintain regular health checkups.",
                         "Follow professional medical advice."]

        return recommendations

    recommendations=["Continue maintaining a healthy lifestyle.",
                     "Attend regular health screenings.",
                     "Maintain regular physical activity."]

    return recommendations


# Disclaimer
def generate_disclaimer():
    """ Generate system disclaimer."""
    return("\nThis AI system is intended only to assist healthcare professionals and"
           " should not be considered as a medical diagnosis.")


# Patient Explanation Layer
def generate_patient_explanation(prediction_label, prediction_confidence, lime_dictionary):
    """ Generate patient explanation. """
    # Generate Explanation
    explanation=(generate_simple_explanation(prediction_label))

    # Important Features
    important_features=(extract_important_features(lime_dictionary))

    # Recommendation
    recommendation=(generate_health_recommendation(prediction_label))

    # Disclaimer
    disclaimer=(generate_disclaimer())

    # Display Results
    print("\n")
    print("="*70)
    print("AI HEALTH REPORT")
    print("="*70)
    print(f"\nPrediction : " f"{prediction_label}")
    print(f"\nPrediction Confidence : " f"{prediction_confidence:.2f}%")
    print("\n")
    print("-"*70)
    print("EXPLANATION")
    print("-"*70)
    print(explanation)


    # Important Features
    print("\n")
    print("-"*70)
    print("IMPORTANT CONTRIBUTORS")
    print("-"*70)

    for feature in (important_features):
        print(f"• {feature}")

    # Recommendations
    print("\n")
    print("-"*70)
    print("HEALTH RECOMMENDATIONS")
    print("-"*70)

    for advice in (recommendation):
        print(f"• {advice}")


    # Disclaimer
    print("\n")
    print("-"*70)
    print("DISCLAIMER")
    print("-"*70)
    print(disclaimer)
    print("\n")
    print("="*70)
    return{"Prediction": prediction_label,
           "Confidence": prediction_confidence,
            "Important Features": important_features,
            "Explanation": explanation,
            "Recommendation": recommendation,
            "Disclaimer": disclaimer}



```

```python
# Algorithm 9
# Medical Decoder Layer

# Medical Terminologies
FEATURE_NAMES = {"Age": "Patient Age",
                 "Sex": "Biological Sex",
                 "Chest Pain": "Type of Chest Pain",
                 "Resting Blood Pressure": "Resting Blood Pressure",
                 "FBS": "Fasting Blood Sugar",
                 "Rest ECG": "Resting Electrocardiographic Results",
                 "Maximum Heart Rate": "Maximum Heart Rate Achieved",
                 "Exang": "Exercise Induced Angina",
                 "Old Peak": "ST Depression Induced by Exercise",
                 "Slope": "Slope of Peak Exercise ST Segment"}

# Medical Description
FEATURE_DESCRIPTION = {"Age": "Increasing age is associated with cardiovascular risk.",
                       "Sex": "Biological sex may influence cardiovascular disease patterns.",
                       "Chest Pain": "Chest pain characteristics are important indicators of cardiovascular abnormalities.",
                       "Resting Blood Pressure": "Elevated resting blood pressure may increase cardiovascular risk.",
                       "FBS": "High fasting blood sugar levels may indicate diabetes related cardiovascular complications.",
                       "Rest ECG": "Electrocardiographic abnormalities may indicate cardiac dysfunction.",
                       "Maximum Heart Rate": "Reduced maximum heart rate during exercise testing may indicate cardiovascular abnormalities.",
                       "Exang": "Chest pain during exercise may indicate reduced blood flow to the heart.",
                       "Old Peak": "ST depression during exercise may indicate myocardial ischemia.",
                       "Slope": "Changes in the ST segment slope may indicate cardiovascular abnormalities."}

# Simple Description
FEATURE_SIMPLE_DESCRIPTION = {"Age": "Your age may affect your heart health.",
                              "Sex": "Some heart conditions occur differently in males and females.",
                              "Chest Pain": "Chest pain may indicate problems related to heart function.",
                              "Resting Blood Pressure": "Blood pressure is an important indicator of heart health.",
                              "FBS": "Blood sugar levels may affect heart health.",
                              "Rest ECG": "The ECG measures the electrical activity of the heart.",
                              "Maximum Heart Rate": "Heart rate measurements may provide information about heart performance.",
                              "Exang": "Chest pain during exercise may indicate increased stress on the heart.",
                              "Old Peak": "Changes observed during exercise may indicate heart related conditions.",
                              "Slope": "ECG changes may provide important information regarding heart health."}


# Decode Features
def decode_feature_name(feature_name):
    """ Decode feature names. """
    if feature_name in FEATURE_NAMES:
        return FEATURE_NAMES[feature_name]
    return feature_name

# Medical Description
def decode_medical_description(feature_name):
    """ Return medical interpretation. """
    if feature_name in (FEATURE_DESCRIPTION):
        return FEATURE_DESCRIPTION[feature_name]
    return ""

# Patient Description
def decode_patient_description(feature_name):
    """ Return simple explanation. """
    if feature_name in (FEATURE_SIMPLE_DESCRIPTION):
        return(FEATURE_SIMPLE_DESCRIPTION[feature_name])
    return ""

# Decode Results
def decode_results( lime_dictionary):
    """ Decode all important features. """
    ranking=(lime_dictionary["Feature Ranking"])
    decoded_results=[]

    for feature,weight in ranking:
        decoded_results.append({"Feature": decode_feature_name(feature),
                                "Medical Description": decode_medical_description(feature),
                                "Simple Description": decode_patient_description(feature)})
    return decoded_results


# Display Results
def display_decoded_results(decoded_results):
    print("\n")
    print("="*70)
    print("MEDICAL KNOWLEDGE REPORT")
    print("="*70)

    for result in decoded_results:
        print("\n")
        print(f"Feature : " f"{result['Feature']}")
        print(f"\nMedical Description : " f"{result['Medical Description']}")
        print(f"\nSimple Description : " f"{result['Simple Description']}")
        print("\n")
        print("-"*70)
```

```python
# Algorithm 10
# Display and Save Results Layer

import json
import os
import pandas as pd

# Display Results
def display_results(prediction_label, prediction_confidence, doctor_results, patient_results, lime_dictionary, decoded_results):
    """ Display all explanation results. """
    print("\n")
    print("="*70)
    print("HEART DISEASE EXPLANATION REPORT")
    print("="*70)
    print(f"\nPrediction : " f"{prediction_label}")
    print(f"\nPrediction Confidence : " f"{prediction_confidence:.2f}%")

    print("\n")
    print("="*70)
    print("REPORT GENERATED SUCCESSFULLY")
    print("="*70)


# Save Doctor Report
def save_doctor_report(doctor_results, file_name="Doctor_Report.txt"):
    with open(file_name, "w", encoding="utf-8") as file:
        for key,value in (doctor_results.items()):
            file.write(f"{key}\n")
            file.write(f"{value}\n\n")

# Save Patient Report
def save_patient_report(patient_results, file_name="Patient_Report.txt"):
    with open(file_name, "w", encoding="utf-8") as file:
        for key,value in (patient_results.items()):
            file.write(f"{key}\n")
            file.write(f"{value}\n\n")

# Save LIME Results
def save_lime_results(lime_dictionary, file_name="LIME_Report.json"):
    with open(file_name, "w") as file:
        json.dump(lime_dictionary, file, indent=4)

# Save Medical Knowledge Results
def save_medical_results(decoded_results, file_name= "Medical_Report.csv"):
    dataframe = pd.DataFrame(decoded_results)
    dataframe.to_csv(file_name, index=False)

# Save Complete Results
def save_complete_results(doctor_results, patient_results, lime_dictionary, decoded_results):
    """ Save every generated report. """
    save_doctor_report(doctor_results)
    save_patient_report(patient_results)
    save_lime_results(lime_dictionary)
    save_medical_results(decoded_results)
    print("\n")
    print("="*70)
    print("ALL REPORTS SAVED SUCCESSFULLY")
    print("="*70)

# Execute Algorithm 10
def execute_algorithm_10(prediction_label, prediction_confidence, doctor_results, patient_results, lime_dictionary, decoded_results):
    display_results(prediction_label, prediction_confidence, doctor_results, patient_results, lime_dictionary, decoded_results)
    save_complete_results(doctor_results, patient_results, lime_dictionary,  decoded_results)
```

```python
# Display Results
def display_results(prediction_label, prediction_confidence, probability_scores, lime_dictionary, doctor_results, patient_results, decoded_results):
    """ Display complete explanation results. """
    print("\n")
    print("="*80)
    print("HEART DISEASE EXPLANATION REPORT")
    print("="*80)

    # Prediction Results
    print("\nPrediction Result")
    print("-"*80)
    print(f"Prediction : " f"{prediction_label}")
    print(f"Prediction Confidence : " f"{prediction_confidence:.2f}%")
    print(f"\nNo Heart Disease : " f"{probability_scores[0]:.2f}%")
    print(f"Heart Disease : " f"{probability_scores[1]:.2f}%")

    # Positive Contributors
    print("\n")
    print("-"*80)
    print("POSITIVE CONTRIBUTORS")
    print("-"*80)

    for feature,weight in (lime_dictionary["Positive Contributors"]:
        print(f"{feature}")


    # Negative Contributors
    print("\n")
    print("-"*80)
    print("NEGATIVE CONTRIBUTORS")
    print("-"*80)

    for feature,weight in (lime_dictionary["Negative Contributors"]):
        print(f"{feature}")

    # Feature Ranking
    print("\n")
    print("-"*80)
    print("FEATURE RANKING")
    print("-"*80)

    rank = 1

    for feature,weight in (lime_dictionary["Feature Ranking"]):
        print(f"{rank}) {feature}")
        rank +=1

    # Doctor's Report
    print("\n")
    print("-"*80)
    print("DOCTOR'S REPORT")
    print("-"*80)
    print(doctor_results["Interpretation"])
    print(doctor_results["Recommendation"])

    # Patient's Report
    print("\n")
    print("-"*80)
    print("PATIENT'S REPORT")
    print("-"*80)
    print(patient_results["Explanation"])

    # Medical Knowledge
    print("\n")
    print("-"*80)
    print("MEDICAL KNOWLEDGE")
    print("-"*80)

    for result in decoded_results:
        print(f"\n{result['Feature']}")
        print(result["Medical Description"])

    print("\n")
    print("="*80)
    print("REPORT GENERATED SUCCESSFULLY")
    print("="*80)
```

```python
# Save Results
import os
import json
import pandas as pd
from datetime import datetime

# Save Results
def save_results(prediction_label, prediction_confidence, probability_scores, doctor_results, patient_results, lime_dictionary,  decoded_results):
    """ Save complete explanation results. """

    # Create Folder
    os.makedirs("Results", exist_ok=True)

    # Time Stamp
    current_time = (datetime.now().strftime("%Y_%m_%d_%H_%M_%S"))

    # Prediction Results
    prediction_results={"Prediction": prediction_label,
                        "Prediction Confidence": prediction_confidence,
                        "No Heart Disease": probability_scores[0],
                        "Heart Disease": probability_scores[1]}


    with open(f"Results/Prediction_{current_time}.json", "w") as file:
        json.dump(prediction_results, file, indent=4)

    # Doctor Results
    with open(f"Results/Doctor_{current_time}.txt", "w", encoding="utf-8" ) as file:
        for key,value in (doctor_results.items()):
            file.write(f"{key}\n")
            file.write(f"{value}\n\n")

    # Patient Results
    with open(f"Results/Patient_{current_time}.txt", "w", encoding="utf-8") as file:
        for key,value in (patient_results.items()):
            file.write(f"{key}\n")
            file.write(f"{value}\n\n")

    # LIME Results
    with open(f"Results/LIME_{current_time}.json", "w") as file:
        json.dump(lime_dictionary, file, indent=4, default=str)

    # Medical Knowledge Results
    dataframe = pd.DataFrame(decoded_results)
    dataframe.to_csv(f"Results/Medical_" f"{current_time}.csv", index=False)

    # Display Results
    print("\n")
    print("="*80)
    print("ALL REPORTS SAVED SUCCESSFULLY")
    print("\nSaved Files")
    print(f"\nPrediction_{current_time}.json")
    print(f"Doctor_{current_time}.txt")
    print(f"Patient_{current_time}.txt")
    print(f"LIME_{current_time}.json")
    print(f"Medical_{current_time}.csv")
    print("="*80)
```
