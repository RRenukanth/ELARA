# Balance Dataset

### 1. Import Libraries and Load Dataset

```python
import pandas as pd
from datetime import datetime

dataset = pd.read_csv("preprocessed_dataset_2026-07-04_07-18-06.csv")
```

### 2. Count class distribution

```python
print("Original Class Distribution:")
print(dataset['target'].value_counts())
```

### 3. Separate classes

```python
class_0 = dataset[dataset['target'] == 0]
class_1 = dataset[dataset['target'] == 1]
```

### 4. Find minority class size

```python
min_samples = min(len(class_0), len(class_1))
```

### 5. Randomly sample equal number from both classes

```python
class_0 = class_0.sample(n=min_samples, random_state=42)
class_1 = class_1.sample(n=min_samples, random_state=42)
```

### 6. Combine both classes

```python
balanced_dataset = pd.concat([class_0, class_1])
```


### 7. Shuffle dataset

```python
balanced_dataset = balanced_dataset.sample(frac=1, random_state=42).reset_index(drop=True)

print("\nBalanced Class Distribution:")
print(balanced_dataset['target'].value_counts())

print("\nBalanced Dataset Shape:")
print(balanced_dataset.shape)
```

### 8. Save the dataset

```python
# Get current date and time
current_time = datetime.now().strftime("%Y-%m-%d_%H-%M-%S")

# Create filename
filename = f"preprocessed_dataset_balanced_{current_time}.csv"

# Save balanced dataset
balanced_dataset.to_csv(filename, index=False)

print(f"Balanced dataset saved successfully as: {filename}")
```
