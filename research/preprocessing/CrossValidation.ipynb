# K-Fold Cross Validation 

### 1. Import libraries and load the dataset

```python
from sklearn.model_selection import StratifiedKFold
import pandas as pd
import pickle
balanced_dataset = pd.read_csv("preprocessed_dataset_balanced_2026-07-04_07-32-05.csv")
```

### 2. Separate Features and Target

```python
X = balanced_dataset.drop('target', axis=1)
y = balanced_dataset['target']
```

### 3. 5-Fold Cross Validation (80/20 Split)

```python
kfold = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

fold = 1

for train_index, test_index in kfold.split(X, y):

    print("=" * 50)
    print(f"Fold {fold}")

    X_train = X.iloc[train_index]
    X_test = X.iloc[test_index]

    y_train = y.iloc[train_index]
    y_test = y.iloc[test_index]

    print("Training Samples :", len(X_train))
    print("Testing Samples  :", len(X_test))

    print("\nTraining Class Distribution")
    print(y_train.value_counts())

    print("\nTesting Class Distribution")
    print(y_test.value_counts())

    fold += 1
```

### 4. Save the Folds

```python
X = balanced_dataset.drop('target', axis=1)
y = balanced_dataset['target']

kfold = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

folds = []

for train_idx, test_idx in kfold.split(X, y):
    folds.append({
        "train_idx": train_idx,
        "test_idx": test_idx
    })

# Save folds
with open("cross_validation_folds.pkl", "wb") as f:
    pickle.dump(folds, f)

print("Cross-validation folds saved successfully.")
```
