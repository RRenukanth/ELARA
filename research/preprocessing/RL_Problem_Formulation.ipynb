# RL Problem Formulation from Dataset



### 1. Import Libraries

```python
import pandas as pd
import numpy as np
import pickle
```

### 3. Load Preprocessed Dataset and Folds

```python
# Load balanced dataset
dataset = pd.read_csv("preprocessed_dataset_2026-07-04_07-18-06.csv")
import pickle
import pandas as pd

# Load folds
with open("cross_validation_folds.pkl", "rb") as f:
    folds = pickle.load(f)
```

### 4. Separate Features and Target

```python
X = dataset.drop("target", axis=1)
y = dataset["target"]

print("Dataset loaded successfully.")
print("Total Samples:", len(dataset))
```

### 5. RL Formulation for each Fold

```python
for fold_number, fold in enumerate(folds, start=1):

    print("=" * 60)
    print(f"Fold {fold_number}")
    print("=" * 60)

    train_idx = fold["train_idx"]
    test_idx = fold["test_idx"]

    # Training data
    X_train = X.iloc[train_idx]
    y_train = y.iloc[train_idx]

    # Testing data
    X_test = X.iloc[test_idx]
    y_test = y.iloc[test_idx]

   
    # RL State Space
    states = X_train.values

    state_size = states.shape[1]

    print("Training Samples :", len(X_train))
    print("Testing Samples  :", len(X_test))

    print("State Shape :", states.shape)
    print("State Dimension :", state_size)

    # Action Space
    actions = [0, 1]

    action_size = len(actions)

    print("Action Space :", actions)


    # Environment Information
    print("\n========== RL Environment ==========")

    print("Number of States :", len(states))
    print("Number of Features :", state_size)
    print("Number of Actions :", action_size)

    print("----------------------------------------")
```

### 6. Save the RL Formulation

```python
rl_environment = []

for fold_number, fold in enumerate(folds, start=1):

    train_idx = fold["train_idx"]
    test_idx = fold["test_idx"]

    X_train = X.iloc[train_idx]
    X_test = X.iloc[test_idx]

    y_train = y.iloc[train_idx]
    y_test = y.iloc[test_idx]

    environment = {

        "fold": fold_number,

        "states_train": X_train.values,

        "labels_train": y_train.values,

        "states_test": X_test.values,

        "labels_test": y_test.values,

        "state_size": X_train.shape[1],

        "action_size": 2
    }

    rl_environment.append(environment)

# Save RL environment
with open("rl_environment.pkl", "wb") as file:

    pickle.dump(rl_environment, file)

print("RL Environment saved successfully.")
```
